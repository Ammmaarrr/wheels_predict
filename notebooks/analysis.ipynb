# PakWheels Car Listings Data Cleaning & Validation

This notebook will guide you through loading, inspecting, cleaning, and flattening your PakWheels car listings data for analysis.

## 1. Import Required Libraries

We'll use pandas for data manipulation and numpy for numerical operations.

```python
import pandas as pd
import numpy as np
import json
```

## 2. Load the Data

Read the car listings CSV file into a pandas DataFrame.

```python
csv_path = 'data/pakwheels_listings.csv'
df = pd.read_csv(csv_path)
print("Shape:", df.shape)
df.head()
```

## 3. Inspect DataFrame Structure

Check the columns, data types, and a preview of the data.

```python
print("Columns:", df.columns.tolist())
df.info()
df.head()
```

## 4. Check for Missing Values

See how many missing values are present in each column.

```python
df.isnull().sum()
```

## 5. Handle Missing Data

Fill missing numeric values with the median and string/object columns with empty strings.

```python
df_filled = df.copy()
for col in df_filled.select_dtypes(include=[np.number]).columns:
    df_filled[col] = df_filled[col].fillna(df_filled[col].median())
for col in df_filled.select_dtypes(include='object').columns:
    df_filled[col] = df_filled[col].fillna('')
df_filled.isnull().sum()
```

## 6. Remove Duplicates

Drop duplicate rows to ensure data quality.

```python
df_cleaned = df_filled.drop_duplicates()
print("Rows after cleaning:", df_cleaned.shape[0])
```

## 7. Validate and Convert Data Types

Convert columns like 'price' to numeric if needed.

```python
if 'price' in df_cleaned.columns:
    df_cleaned['price'] = pd.to_numeric(df_cleaned['price'], errors='coerce')
df_cleaned.dtypes
```

## 8. Clean and Standardize Column Names

Make column names lowercase, strip whitespace, and replace spaces with underscores.

```python
df_cleaned.columns = df_cleaned.columns.str.strip().str.lower().str.replace(' ', '_')
for col in df_cleaned.select_dtypes(include='object').columns:
    df_cleaned[col] = df_cleaned[col].str.strip()
df_cleaned.head()
```

## 9. Save Cleaned Data

Export the cleaned DataFrame to a new CSV file.

```python
df_cleaned.to_csv('data/pakwheels_listings_cleaned.csv', index=False)
```

## 10. Extract and Flatten Car Details from the `data` Column

If your car details are stored as JSON or dict in the `data` column, flatten them for analysis.

```python
def ensure_dict(val):
    if isinstance(val, str):
        try:
            return json.loads(val)
        except Exception:
            return {}
    elif isinstance(val, dict):
        return val
    else:
        return {}

if 'data' in df_cleaned.columns:
    df_cleaned['data'] = df_cleaned['data'].apply(ensure_dict)
    details_df = pd.DataFrame(df_cleaned['data'].tolist())
    # Optionally, join with other columns (like id, car_url)
    if 'id' in df_cleaned.columns:
        details_df['id'] = df_cleaned['id']
    if 'car_url' in df_cleaned.columns:
        details_df['car_url'] = df_cleaned['car_url']
    details_df.head()
else:
    print("No 'data' column found in the DataFrame.")
```

Display the first few rows and columns of the normalized car details DataFrame.

```python
print('Columns:', details_df.columns.tolist())
details_df.head(10)
```
